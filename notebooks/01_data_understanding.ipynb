```python
def convert_area_to_marla(value):
    value= str(value).strip()

    number = float(value.split()[0])
    unit = value.split()[1].lower()

    if unit == "marla":
        return number
    elif unit == "kanal":
        return number*20
    else:
        None

df["area_marla"] = df["area"].apply(convert_area_to_marla)
```
