```python
def convert_area_to_marla(value):
    value = str(value).strip().replace(",", "")

    parts = value.split()

    number = float(parts[0])
    unit = parts[1].lower()

    if unit == "marla":
        return number
    elif unit == "kanal":
        return number * 20
    else:
        return None


df["area_marla"] = df["area"].apply(convert_area_to_marla)
```
